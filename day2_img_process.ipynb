```python
import cv2
import matplotlib.pyplot as plt
%matplotlib inline
```

## 图像处理

```python
img = cv2.imread('./assets/1.jpg')

ret, thresh1 = cv2.threshold(img, 127, 255, cv2.THRESH_BINARY) # 亮点全白，暗点全黑
ret, thresh2 = cv2.threshold(img, 127, 255, cv2.THRESH_BINARY_INV) # inverse
ret, thresh3 = cv2.threshold(img, 127, 255, cv2.THRESH_TRUNC) # 大于 127 = 127
ret, thresh4 = cv2.threshold(img, 127, 255, cv2.THRESH_TOZERO) # 小于127 = toZero
ret, thresh5 = cv2.threshold(img, 127, 255, cv2.THRESH_TOZERO_INV) # 大于 127 = 0

plt.subplot(231), plt.imshow(img), plt.title("origion"), plt.axis('off')
plt.subplot(232), plt.imshow(thresh1), plt.title("binary"), plt.axis('off')
plt.subplot(233), plt.imshow(thresh2), plt.title("inv_b"), plt.axis('off')
plt.subplot(234), plt.imshow(thresh3), plt.title("trunc"), plt.axis('off')
plt.subplot(235), plt.imshow(thresh4), plt.title("toZero"), plt.axis('off')
plt.subplot(236), plt.imshow(thresh5), plt.title("inv_t"), plt.axis('off')
```

### 图像平滑

```python
img = cv2.imread('./assets/1.jpg')

# 均值滤波（平均卷积）
blur = cv2.blur(img, (3, 3)) # 3x3 中心像素点 = 均值
# 方框滤波，同上，归一化（normalize=False 时，取和，越界当 255 处理）
box = cv2.boxFilter(img, -1, (3 ,3), normalize=True)
# 高斯滤波，离得近的权重大，1 = 高斯核的标准差 (sigma) 参数，越大越模糊
gussian = cv2.GaussianBlur(img, (7, 7), 7)
median = cv2.medianBlur(img, 5)

plt.subplot(231), plt.imshow(img), plt.title("origion"), plt.axis('off')
plt.subplot(232), plt.imshow(blur), plt.title("blur"), plt.axis('off')
plt.subplot(233), plt.imshow(box), plt.title("box"), plt.axis('off')
plt.subplot(234), plt.imshow(gussian), plt.title("gussian"), plt.axis('off')
plt.subplot(235), plt.imshow(median), plt.title("median"), plt.axis('off')
```
